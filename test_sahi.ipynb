```python
from sahi import AutoDetectionModel
from sahi.utils.cv import read_image
from sahi.utils.file import download_from_url
from sahi.predict import get_prediction, get_sliced_prediction, predict
from IPython.display import Image
import os

from pathlib import Path
detection_model = AutoDetectionModel.from_pretrained(
    model_type='ultralytics',
    model_path="out/study1/thread_1/yolo11n_64/iter_03/weights/best.pt",
    confidence_threshold=0.25,
    device="cuda:0", # or 'cuda:0' if GPU is available
)
```

```python
dir_images = "/home/niche/data/finding-ants/study1/test_b03/images/"
path_images = list(Path(dir_images).glob("*.jpg"))
path_images.sort()
print(path_images[0])
from PIL import Image
Image.open(path_images[4]).show()
```

```python
result = get_sliced_prediction(
    Image.open(path_images[4]),
    detection_model,
    slice_height = 256,
    slice_width = 256,
    overlap_height_ratio = 0.2,
    overlap_width_ratio = 0.2
)
```
