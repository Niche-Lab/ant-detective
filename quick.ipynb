```python
import pandas as pd
# ram
data = pd.read_csv("out/s2_raw.csv").loc[:, ["model", "cat", "strategy", "max_mem"]]
data = data.query("strategy != 'baseline'")
# aggregate
data.head()
```

```python
data_gr = data.groupby(["model",]).agg(
    mean_mem=("max_mem", "mean"),
    std_mem=("max_mem", "std"),
).reset_index()
data_gr["mean_mem"] = round(data_gr["mean_mem"] / 1024, 2)  # convert to GB
data_gr["std_mem"] = round(data_gr["std_mem"] / 1024, 2)  # convert to GB
data_gr
```
